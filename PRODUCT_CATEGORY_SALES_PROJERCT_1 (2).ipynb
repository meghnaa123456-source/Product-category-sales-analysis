**Data quality checking**

```python
display (df.duplicated().sum())
display (df.drop_duplicates(inplace=True))
display (df.isna().sum())
```

**DATA CLEANING:**

```python
df = df.copy()
```

```python
df.columns = df.columns.str.strip()
```

```python
df['Date']=pd.to_datetime(df['Date'],errors='coerce')
```

```python
df=df.dropna(subset=['Date'])
```

```python
df['Date'].isnull().sum()
```

```python
df['Category']=df['Category'].fillna(df['Category'].mode()[0])
```

```python
df['Category'] = df['Category'].astype(str)
```

```python
df['Category'].isnull().sum()
```

```python
df['Product']=df['Product'].fillna(df['Product'].mode()[0])
```

```python
df['Product']=df['Product'].astype(str)
```

```python
df['Product'].isnull().sum()
```

```python
df['Gender']=df['Gender'].fillna(df['Gender'].mode()[0])
```

```python
df['Gender']=df['Gender'].astype(str)
```

```python
df['Gender'].isnull().sum()
```

```python
df['Quantity']=pd.to_numeric(df['Quantity'],errors='coerce')
df['Quantity']=df['Quantity'].fillna(df['Quantity'].mode()[0])
```

```python
df['Quantity']=df['Quantity'].astype('int')
```

```python
df['Quantity'].isnull().sum()
```

```python
df['Unit_Price_Rs']=pd.to_numeric(df['Unit_Price_Rs'],errors='coerce')
df['Unit_Price_Rs']=df['Unit_Price_Rs'].fillna(df['Unit_Price_Rs'].median())
```

```python
df['Unit_Price_Rs'] = df['Unit_Price_Rs'].fillna(0).astype('int')
```

```python
df['Unit_Price_Rs'].isnull().sum()
```

```python
df['Payment']=df['Payment'].fillna(df['Payment'].mode()[0])
```

```python
df['Payment'].isnull().sum()
```

```python
display (df.isna().sum())
```

```python
text_cols =['Category','Product','Gender','Payment']
for col in text_cols:
  df[col]=(
      df[col].astype("string")
         .str.strip()
         .str.title()
  )
  print(df[text_cols].head())
```

# **Visualizations: Univariate Analysis**

**HISTOGRAM**

```python
plt.figure(figsize=(8,4))
sns.histplot(df['Total_Sales_Rs'],bins=25,kde=True)
plt.title("distribution of Transaction sales")
plt.xlabel("Total sales(Rs)")
plt.ylabel("Frequency")
plt.show()
```

**COUNT PLOT : CATEGORY**

```python
plt.figure(figsize=(8,5))
sns.countplot(data=df,x='Category')
plt.title("Category Count")
plt.xlabel("Category")
plt.ylabel("Count")
plt.show()
```

 **COUNT PLOT : PAYMENT**

```python
plt.figure(figsize=(8,5))
sns.countplot(data=df,x="Payment")
plt.title("Payment method Count")
plt.xlabel("payment method")
plt.ylabel("Count")
plt.show()
```

# **Bivariate Analysis**

**BOX PLOT : CATEGORY-WISE SALES**

```python
plt.figure(figsize=(8,5))
sns.boxplot(data=df, x='Category', y='Total_Sales_Rs')
plt.title('Sales Distribution by Category ')
plt.xlabel('Category')
plt.ylabel('Total Sales (Rs)')
plt.show()
```

**LINE PLOT : MONTHLY SALES TREND**

```python
df['Date'] = pd.to_datetime(df['Date'], errors='coerce')
monthly=df.groupby(df['Date'].dt.month)['Total_Sales_Rs'].sum()
monthly.plot(kind='line', marker='o',figsize=(8,5))
plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Sales (Rs)')
plt.show()
```

# **Multivariate Analysis**

**HEATMAP : CORRELATION**

```python
plt.figure(figsize=(8,5))
sns.heatmap(
    df[['Quantity','Unit_Price_Rs','Total_Sales_Rs']].corr(),
    annot=True,
    cmap='coolwarm'
)
plt.title('Sales Correlation Heatmap')
plt.show()
```

**INSIGHTS**



1.  Most transactions have lower sales values,while a few transactions have higher sales values.

2.  Food category has a higher number of transaction compared to Beverages.

3.  Customers use differsnt payment methods such as Cash,UPI and Card.

4.  The box plot shows some outliers in both Food and Beverages Categories.

5.  The scatter plot shows that total sales generally increase when the quantity increase.

6.  Monthly sales shows variations throughout the year.

7.   The heatmap shows a positive relationship between  Unit Price and Total    Sales.
